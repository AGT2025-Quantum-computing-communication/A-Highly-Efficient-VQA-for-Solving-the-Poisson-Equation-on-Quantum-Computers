```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
RMSE_2q = [0.004959513864746268, 0.001341496398741327, 0.00036273175973551526, 0.00019044123484192497]
RMSE_3q = [0.025268870089557714, 0.009712802814407678, 0.0028662668362718375, 0.0009667967818177185]
RMSE_4q = [0.09967562051294877, 0.02775565170847619, 0.025778189742145685, 0.019660281072207023]
RMSE = np.array([RMSE_2q, RMSE_3q, RMSE_4q])
```

```python
groups = ['2q', '3q', '4q']

data = RMSE

# X positions for groups
x = np.arange(len(groups))

fig, ax = plt.subplots(figsize=(10, 6))
color = ['blue','red','green','black']
# Plot bars for each group
for i, group_data in enumerate(data):
    offsets = np.linspace(-0.3, 0.3, len(group_data))
    for j, value in enumerate(group_data):
        ax.bar(x[i] + offsets[j], value, color = color[j], width=0.1)
        
# ax.bar_label(ax.containers[0], label_type='edge', color='red', rotation=90, fontsize=7, padding=3)
# Labels
ax.set_xticks(x)
ax.set_xticklabels(groups)
ax.set_ylabel('RMSE',fontsize=10,weight='bold')
# ax.set_title('Replicated Sketch Chart')
import matplotlib.ticker as ticker
formatter = ticker.ScalarFormatter(useMathText = True)
formatter.set_powerlimits((-3,-3))
ax.yaxis.set_major_formatter(formatter)
ax.set_xticks(x,['2 qubits', '3 qubits', '4 qubits'], rotation = 30)
# legend=plt.legend(fontsize=20,prop=legend_properties)
ax = plt.gca()
for tick in ax.get_xticklabels():
	tick.set_fontweight('bold')
for tick in ax.get_yticklabels():
	tick.set_fontweight('bold')

x1 = np.arange(len([1]))
inset_ax = fig.add_axes([0.22, 0.4, 0.2, 0.2])  # [left, bottom, width, height]
offsets = np.linspace(-0.3, 0.3, len(RMSE[0]))
for j, value in enumerate(RMSE[0]):
    inset_ax.bar(RMSE[0] + offsets[j], value, color = color[j], width = 0.1)
inset_ax.set_xticks(x1,['2 qubits'])   # Zoom into part of the data

ax = plt.gca()
for tick in ax.get_xticklabels():
	tick.set_fontweight('bold')
for tick in ax.get_yticklabels():
	tick.set_fontweight('bold')
# inset_ax.set_ylim(0.999, 1.0005)
plt.savefig('RMSE.pdf')
```
